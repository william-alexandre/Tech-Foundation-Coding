```python
#importar NumPy
import numpy as np
```

```python
# Definir a matriz de Entrada
entradas = np.array([[0,0],
                     [0,1],
                     [1,0],
                     [1,1]])
entradas
```

```python
#Definir a Classe Verdade
#saidas = np.array([0,0,0,1]) # Operador AND
#saidas = np.array([0,1,1,1]) # Operador OR
saidas = np.array([0,1,1,0]) # Operador XOR
saidas
```

```python
# Definir os pesos iniciais
pesos = np.array([0.0,0.0])
pesos
```

```python
# Definir a variável taxaAprendizagem
taxaAprendizagem = 0.1
```

```python
# Definir a Função de Ativação Step Function
def stepFunction(soma):
    if (soma >= 1):
        return 1
    return 0
```

```python
# Definir a função Soma
def somaFunction(entradas):
  #Multiplicar os pesos
  return entradas.dot(pesos)
```

```python
# Definindo a Função de treino do Modelo
def treinarModelo():
  erroTotal = 1
  while (erroTotal != 0):
    erroTotal = 0
    for i in range(len(saidas)):

      #Chamada da Funçaõ soma para aplicação dos pesos
      soma = somaFunction(entradas[i])

      #Chamada da Função de Ativação
      saidaCalculada = stepFunction(soma)

      #Calcular o erro
      erro = abs(saidas[i] - saidaCalculada)

      #Adicionar o resultado na variavel erroTotal
      erroTotal += erro

      #Realizar os ajustes nos pesos
      for j in range(len(pesos)):

        #FÓRMULA DE ATUALIZAÇÃO DOS PESOS
        # peso(n + 1) = peso(n) + (taxaAprendizagem * entrada * erro)
        pesos[j] = pesos[j] + (taxaAprendizagem * entradas[i][j] * erro)

       #Peso Atualizado
        print('Peso atualizado: ' + str(pesos[j]))

      print ('Total de erros: '+ str(erroTotal))
```

```python
# Executar o treino do mode
treinarModelo()
print('Rede Neural Treinada')
print(stepFunction(somaFunction(entradas[0])))
print(stepFunction(somaFunction(entradas[1])))
print(stepFunction(somaFunction(entradas[2])))
print(stepFunction(somaFunction(entradas[3])))
```
